```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from udrud_framework import cookes_base_estimate, cluster_jackknife
from gen_weibull import generate_3p_weibull

# ==========================================
# REPRODUCIBILITY: Set the global random seed
# ==========================================
np.random.seed(2026) # You can choose any integer (e.g., the year of your submission)

# Simulation Parameters
true_gamma = -5000
scale_param = 1000 # lambda
shape_params = [1.0, 1.5, 2.0, 2.5, 3.0] # Test various curvatures
n_sims = 1000
n_obs = 1000

for alpha in shape_params:
    print(f"\nTesting Weibull Alpha = {alpha}")
    results = {"Cooke-Base": [], "Cluster-Jackknife": []}

    for _ in range(n_sims):
        # Generate data
        y = generate_3p_weibull(n_obs, true_gamma, scale_param, alpha)
        y = np.sort(y)
        y_tail = y[:50] # Take the lowest 50 points

        results["Cooke-Base"].append(cookes_base_estimate(y_tail))
        results["Cluster-Jackknife"].append(cluster_jackknife(y_tail))

    # Calculate MSE and Bias
    for method, vals in results.items():
        mse = np.mean((np.array(vals) - true_gamma)**2)
        bias = np.mean(np.array(vals) - true_gamma)
        print(f"{method}: MSE = {mse:.4f}, Bias = {bias:.4f}")
```
